## Top Players by Total Points

sort through all players, select top 20 by total points.

This provides a snapshot of the best performing (FPL wise) players of the season.

```python
# print the datatype of each column
print(elements_teams.dtypes.to_string())
```

```python
# Get standard descriptive statistics for total points
elements_teams["total_points"].describe()
```

```python
# make a clean, defragmented copy of the DataFrame
elements_teams = elements_teams.copy()

# create a dictionary to translate element_type values into positions
position_map = {
    1: "GK",
    2: "DEF",
    3: "MID",
    4: "FWD"
}

# add position column
elements_teams["position"] = elements_teams["element_type"].map(position_map)

# add price column
elements_teams["price"] = elements_teams["now_cost"] / 10

# top 20 players by total points
elements_teams.sort_values("total_points", ascending=False)[
    ["web_name", "team_name", "position", "price", "selected_by_percent", "total_points"]
].head(20)
```
