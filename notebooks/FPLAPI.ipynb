# Fantasy Premier League API

Exploring the Fantasy Premier League API to view the data available.

Contains information on players, teams, gameweeks, fixtures and more.

```python
# imports
import requests
import pandas as pd
```

# Bootstrap Static Endpoint

```python
# base url for FPL API
url = 'https://fantasy.premierleague.com/api/'

# get request for boostrap-static endpoint
r = requests.get(url + 'bootstrap-static/').json()
```

```python
# return keys of the dictionary from r
r.keys()
```

```python
# lists chips available to FPL managers and the windows (GWs) they can be used
chips_df = pd.DataFrame(r['chips'])
chips_df
```

```python
# list of all gws, deadlines and stats --> events = GWs
events_df = pd.DataFrame(r['events'])
events_df.head()
```

```python
# which gameweeks are in which months
phases_df = pd.DataFrame(r['phases'])
phases_df
```

```python
# all 20 Premier League teams + stats
teams_df = pd.DataFrame(r['teams'])
teams_df
```

```python
# total count of FPL Managers
total_players = r['total_players']
print(total_players)
```

```python
# name and label of element stats
element_stats_df = pd.DataFrame(r['element_stats'])
element_stats_df
```

```python
# types of elements e.g. MID, DEF
# contains the upper and lower bounds of position types for team selection
element_types_df = pd.DataFrame(r['element_types'])
element_types_df
```

```python
# list of all players and stats (summed to current gw when run)
elements_df = pd.DataFrame(r['elements'])
elements_df
```

# Element Summary Endpoint

```python
# get player e.g. player 1
# element-summary/{{player_id}}/ --> replace player_id with value
r = requests.get(url + 'element-summary/1/').json()
r.keys()
```

```python
# players upcoming fixtures for live season
element_fix_df = pd.DataFrame(r['fixtures'])
element_fix_df
```

```python
# player stats of previous games in current season
element_history_df = pd.DataFrame(r['history'])
element_history_df.head()
```

```python
# players stats from previous seasons (summary)
element_history_past_df = pd.DataFrame(r['history_past'])
element_history_past_df
```

# Gameweek Fixtures Endpoint

```python
# fixtures/?event={{gw_id}} --> replace gw_id with value
# event = 1 is the first gw of the season
r = requests.get(url + 'fixtures/?event=1').json()
gw_fixture_df = pd.DataFrame(r)
gw_fixture_df.head()
```

# Gameweek Live Endpoint

```python
# /{{gw_id}}/live/ --> replace gw_id with value
# list of elements, players live gw performance
r = requests.get(url + 'event/1/live/').json()
gw_live_df = pd.DataFrame(r)
gw_live_df
```

# Save dataframes to csv

Viewing the data in the FPL API is useful, but we may want to save it for later use.

```python
# save elements dataframe to csv, which contains all players and all stats, for viewing
elements_df.to_csv(r'..\data\fromAPI\elements.csv', index=False)
```

```python
# save all 20 Premier League teams and stats to csv
teams_df.to_csv(r'..\data\fromAPI\teams.csv', index=False)
```

# Merge elements and teams

merging these two dataframes allows us to know which player belongs to which team.

```python
# find columns that exist in both files
shared_columns = set(elements_df.columns).intersection(set(teams_df.columns))

# remove the column we plan to merge on
problem_columns = shared_columns - {'team_id'}

# print columns that may cause _x and _y after merging
print(problem_columns)
```

```python
# Rename some overlapping columns to differentiate original source
elements_df = elements_df.rename(columns={
    'id': 'player_id',
    'team': 'team_id',
    'code': 'player_code',
    'form': 'player_form'
})

teams_df = teams_df.rename(columns={
    'id': 'team_id',
    'form': 'team_form',
    'name': 'team_name'
})

# we drop code as elements already has team_code, which is the same thing
teams_df = teams_df.drop(columns = ['code'])

# merge players to teams using the team ID
elements_teams = elements_df.merge(
    teams_df,
    on='team_id',
    how='left'
)

with pd.option_context('display.max_rows', None):
    print(elements_teams.loc[0])

# save as backup/for viewing
elements_teams.to_csv(r'..\data\fromAPI\elements_teams.csv', index=False)
```
