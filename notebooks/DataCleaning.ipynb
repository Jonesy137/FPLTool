# Data Cleaning

Look at datatype and values of every column. Make amendments and check for correctness of the data. This is a crucial step before any analysis, as it ensures that the data is in the correct format and that there are no inconsistencies or errors.

### Observations:

Missing values are just empty cells, needs replacing with consistent missing value.

Date columns such as news_added, team_join_date and birth_date are stored as string types.

ID columns such as player_id are stored as integers, not strings. This may be fine.

These columns are empty:
- squad_number, corners_and_indirect_freekicks_text, direct_freekicks_text, penalties_text, team_form, team_division, link_url

Stored as float 64, but non-missing values are whole numbers (decimals are not needed):
- chance_of_playing_next_round, chance_of_playing_this_round, region, corners_and_indirect_freekicks_order, direct_freekicks_order, penalties_order

now_cost is stored in tenths, so we need to divide by 10 to get the actual cost in millions (55 → 5.5m)

```python
# imports
import pandas as pd
```

```python
# load data
elements_teams = pd.read_csv(r'..\data\fromAPI\elements_teams.csv')
```

```python
elements_teams.shape
```

```python
# print the datatype of each column
print(elements_teams.dtypes.to_string())
```

```python
# show sum of all null columns
print(elements_teams.isnull().sum().to_string())
```

```python
elements_teams.columns = (
    elements_teams.columns
    .str.strip()  # remove leading and trailing whitespace from string columns
)
```

```python
# calculate percentage of missing values
missing = elements_teams.isnull().mean() * 100
print(missing[missing > 0].sort_values(ascending=False))
```

```python
# number of duplicate rows
n_dupes = elements_teams.duplicated().sum()
n_dupes
```
